### Depth-First-Search (DFS)

```python
def dfs(graph, start_node):
  """
  Performs a Depth-First Search (DFS) on a graph.

  Args:
    graph: A dictionary representing the graph where keys are nodes and values
           are lists of their neighbors.
    start_node: The node to start the DFS from.

  Returns:
    A list of nodes in the order they were visited during the DFS.
  """

  visited = set()
  traversal_order = []

  def dfs_recursive(node):
    visited.add(node)
    traversal_order.append(node)

    for neighbor in graph.get(node, []):
      if neighbor not in visited:
        dfs_recursive(neighbor)

  dfs_recursive(start_node)
  return traversal_order
```

### Breadth-First Search (BFS)

```python
from collections import deque

def bfs(graph, start_node):
  """
  Performs a Breadth-First Search (BFS) on a graph.

  Args:
    graph: A dictionary representing the graph where keys are nodes and values
           are lists of their neighbors.
    start_node: The node to start the BFS from.

  Returns:
    A list of nodes in the order they were visited during the BFS.
  """

  visited = set()
  traversal_order = []
  queue = deque([start_node])
  visited.add(start_node)

  while queue:
    node = queue.popleft()
    traversal_order.append(node)

    for neighbor in graph.get(node, []):
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)

  return traversal_order
```

### Exercise 02: Define a goal state and print the shortest path to the goal
Complete DFS and BFS implementation code:
* Define a goal state
* Get and print the shortest path to the goal

```python
from collections import deque

def dfs_shortest_path(graph, start_node, goal_node):
  """
  Performs a Depth-First Search (DFS) on a graph and finds the shortest path to the goal state.

  Args:
    graph: A dictionary representing the graph where keys are nodes and values
           are lists of their neighbors.
    start_node: The node to start the DFS from.
    goal_node: The goal node to find the shortest path to.

  Returns:
    A tuple containing:
      - A list of nodes in the order they were visited during the DFS.
      - A list representing the shortest path from the start node to the goal node,
        or None if no path is found.
  """

  visited = set()
  traversal_order = []
  parent = {}  # Dictionary to store parent nodes for path reconstruction

  def dfs_recursive(node):
    visited.add(node)
    traversal_order.append(node)

    if node == goal_node:
      return True  # Goal found

    for neighbor in graph.get(node, []):
      if neighbor not in visited:
        parent[neighbor] = node  # Store the parent of the neighbor
        if dfs_recursive(neighbor):
          return True  # Goal found through this neighbor

    return False  # Goal not found from this node

  if not dfs_recursive(start_node):
    return traversal_order, None  # No path found

  # Reconstruct the shortest path
  path = [goal_node]
  while path[-1] != start_node:
    path.append(parent[path[-1]])
  path.reverse()

  return traversal_order, path
```

```python
from collections import deque

def bfs_shortest_path(graph, start_node, goal_node):
  """
  Performs a Breadth-First Search (BFS) on a graph and finds the shortest path to the goal state.

  Args:
    graph: A dictionary representing the graph where keys are nodes and values
           are lists of their neighbors.
    start_node: The node to start the BFS from.
    goal_node: The goal node to find the shortest path to.

  Returns:
    A tuple containing:
      - A list of nodes in the order they were visited during the BFS.
      - A list representing the shortest path from the start node to the goal node,
        or None if no path is found.
  """

  visited = set()
  traversal_order = []
  queue = deque([(start_node, [start_node])])  # Queue stores (node, path) tuples
  visited.add(start_node)

  while queue:
    node, path = queue.popleft()                # gets the note and its path
    traversal_order.append(node)

    if node == goal_node:
      return traversal_order, path               # Goal found, return the path

    for neighbor in graph.get(node, []):
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append((neighbor, path+[neighbor])) # add node and its path to queue

  return traversal_order, None                    # No path found
```

### Exercise 03: Implement the Uniform Cost Search algorithm

Based on the DFS code, implement a Uniform Cost Search (UCS) algorithm

```python
# prompt: Update the code below to return:
# 1. The cost
# 2. The shortest path
# 3. the list of nodes visited by order of visit

from collections import deque
import heapq

def ucs(graph, start_node, goal_node):
    """
    Performs a Uniform Cost Search (UCS) on a graph.

    Args:
        graph: A dictionary representing the graph where keys are nodes and values
               are lists of (neighbor, cost) tuples.
        start_node: The node to start the UCS from.
        goal_node: The goal node to find the shortest path to.

    Returns:
        A tuple containing:
            - The cost of the shortest path from the start node to the goal node.
            - A list representing the shortest path from the start node to the goal node,
              or None if no path is found.
            - A list of nodes visited in order.
    """
    visited = set()
    priority_queue = [(0, start_node, [start_node])]  # (cost, node, path)
    visited_nodes = []

    while priority_queue:
        cost, current_node, path = heapq.heappop(priority_queue)
        visited_nodes.append(current_node)

        if current_node == goal_node:
            return cost, path, visited_nodes

        visited.add(current_node)

        for neighbor, edge_cost in graph.get(current_node, []):
            if neighbor not in visited:
                new_cost = cost + edge_cost
                heapq.heappush(priority_queue, (new_cost, neighbor, path + [neighbor]))

    return float('inf'), None, visited_nodes # No path found
```

```python
# Example graph with edge costs
graph_with_costs = {
    'A': [('C', 1), ('C', 3)],
    'B': [('D', 5), ('E', 2)],
    'C': [('F', 4)],
    'D': [('G', 7)],
    'E': [('F', 1)],
    'F': [],
    'G': []
}

start_node = 'A'
goal_node = 'F'

cost, shortest_path, visited_nodes = ucs(graph_with_costs, start_node, goal_node)

print(f"UCS:")
print(f"Cost of shortest path: {cost}")

if shortest_path:
    print(f"Shortest path from {start_node} to {goal_node}: {shortest_path}")
else:
    print(f"No path found from {start_node} to {goal_node}")
print(f"Visited nodes: {visited_nodes}")
```
